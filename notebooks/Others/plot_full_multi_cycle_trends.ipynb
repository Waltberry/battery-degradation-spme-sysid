```python
# Cell 1
from __future__ import annotations

import re
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (10, 5)
plt.rcParams["axes.grid"] = True
plt.rcParams["grid.alpha"] = 0.3
plt.rcParams["lines.linewidth"] = 2

PROJECT_ROOT = Path("/home/onyero.ofuzim/projects/battery-degradation-spme-sysid")
METRICS_DIR = PROJECT_ROOT / "results" / "metrics"
TABLES_DIR = PROJECT_ROOT / "results" / "tables"
OUTPUT_DIR = PROJECT_ROOT / "results" / "figures" / "full_multi_cycle_trends"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("PROJECT_ROOT:", PROJECT_ROOT)
print("METRICS_DIR:", METRICS_DIR)
print("TABLES_DIR:", TABLES_DIR)
print("OUTPUT_DIR:", OUTPUT_DIR)

# Cell 2
WINDOW_RE = re.compile(r"_(\d{3})_(\d{3})\.csv$")


def parse_window_from_filename(path: Path) -> tuple[int, int, int]:
    """
    Extract (start, end, span) from filenames like:
    multi_cycle_metrics_026_075.csv
    multi_cycle_parameter_table_126_175.csv
    """
    m = WINDOW_RE.search(path.name)
    if not m:
        raise ValueError(f"Could not parse cycle window from filename: {path}")
    start = int(m.group(1))
    end = int(m.group(2))
    span = end - start + 1
    return start, end, span


def add_file_metadata(df: pd.DataFrame, source_file: Path) -> pd.DataFrame:
    start, end, span = parse_window_from_filename(source_file)
    out = df.copy()
    out["_source_file"] = source_file.name
    out["_window_start"] = start
    out["_window_end"] = end
    out["_window_span"] = span
    out["_non_null_count"] = out.notna().sum(axis=1)
    return out


def load_csv_group(file_paths: list[Path]) -> pd.DataFrame:
    """
    Load a group of chunk CSV files and concatenate them.
    """
    frames = []
    for path in sorted(file_paths):
        df = pd.read_csv(path)
        df = add_file_metadata(df, path)
        frames.append(df)

    if not frames:
        return pd.DataFrame()

    return pd.concat(frames, ignore_index=True)


def deduplicate_by_cycle(df: pd.DataFrame, cycle_col: str = "cycle_idx") -> pd.DataFrame:
    """
    Because chunk files overlap (example: 001_002 and 001_025),
    keep one row per cycle.

    Preference:
    1. rows from wider chunk windows first
    2. rows with more non-null values
    3. later window end
    """
    if df.empty:
        return df.copy()

    sort_cols = ["_window_span", "_non_null_count", "_window_end", cycle_col]
    ascending = [False, False, False, True]

    out = (
        df.sort_values(sort_cols, ascending=ascending)
          .drop_duplicates(subset=[cycle_col], keep="first")
          .sort_values(cycle_col)
          .reset_index(drop=True)
    )
    return out


def print_df_summary(name: str, df: pd.DataFrame) -> None:
    print(f"\n{name}")
    print("-" * len(name))
    print("shape:", df.shape)
    if not df.empty and "cycle_idx" in df.columns:
        print("cycle range:", int(df["cycle_idx"].min()), "to", int(df["cycle_idx"].max()))
        print("unique cycles:", df["cycle_idx"].nunique())
    print("columns:")
    print(list(df.columns))

# Cell 3
metric_files = sorted(METRICS_DIR.glob("multi_cycle_metrics_*.csv"))
param_files = sorted(TABLES_DIR.glob("multi_cycle_parameter_table_*.csv"))

print("Metric files found:")
for f in metric_files:
    print("  ", f.name)

print("\nParameter table files found:")
for f in param_files:
    print("  ", f.name)

metrics_raw = load_csv_group(metric_files)
params_raw = load_csv_group(param_files)

print_df_summary("metrics_raw", metrics_raw)
print_df_summary("params_raw", params_raw)

# Cell 4
metrics_full = deduplicate_by_cycle(metrics_raw, cycle_col="cycle_idx")
params_full = deduplicate_by_cycle(params_raw, cycle_col="cycle_idx")

print_df_summary("metrics_full", metrics_full)
print_df_summary("params_full", params_full)

display(metrics_full.head())
display(params_full.head())

# Cell 5
meta_cols = ["_source_file", "_window_start", "_window_end", "_window_span", "_non_null_count"]

metrics_merge = metrics_full.drop(columns=[c for c in meta_cols if c in metrics_full.columns], errors="ignore")
params_merge = params_full.drop(columns=[c for c in meta_cols if c in params_full.columns], errors="ignore")

full_df = pd.merge(
    metrics_merge,
    params_merge,
    on="cycle_idx",
    how="outer",
    suffixes=("_metrics", "_params"),
)

# Coalesce duplicate columns if they exist in both
for col in list(full_df.columns):
    if col.endswith("_metrics"):
        base = col[:-8]
        col_metrics = f"{base}_metrics"
        col_params = f"{base}_params"
        if col_params in full_df.columns:
            full_df[base] = full_df[col_metrics].combine_first(full_df[col_params])
            full_df.drop(columns=[col_metrics, col_params], inplace=True)

full_df = full_df.sort_values("cycle_idx").reset_index(drop=True)

print_df_summary("full_df", full_df)
display(full_df.head())

# Cell 6
all_cols = list(full_df.columns)

thetaA_cols = [c for c in all_cols if c.lower().startswith("thetaa")]
thetaB_cols = [c for c in all_cols if c.lower().startswith("thetab")]

print("Theta A columns:")
print(thetaA_cols)

print("\nTheta B columns:")
print(thetaB_cols)

print("\nOther important columns:")
for col in [
    "cycle_idx",
    "final_stage_name",
    "stage2_rmse",
    "stage3b_rmse",
    "R0_stage3b",
    "shape_drift_rmse",
]:
    print(f"{col}: {'YES' if col in full_df.columns else 'NO'}")


# Cell 7
full_csv_path = TABLES_DIR / "full_multi_cycle_trend_data_001_272.csv"
full_df.to_csv(full_csv_path, index=False)
print("Saved merged full dataframe to:", full_csv_path)


# Cell 8
def finalize_plot(title: str, xlabel: str, ylabel: str, save_name: str | None = None):
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.tight_layout()
    if save_name is not None:
        save_path = OUTPUT_DIR / save_name
        plt.savefig(save_path, dpi=200, bbox_inches="tight")
        print("Saved:", save_path)
    plt.show()


def plot_single_metric(
    df: pd.DataFrame,
    y_col: str,
    title: str,
    ylabel: str,
    save_name: str,
    marker: str = "o",
):
    if y_col not in df.columns:
        print(f"[WARN] Column not found: {y_col}")
        return

    plot_df = df[["cycle_idx", y_col]].copy()
    plot_df = plot_df.dropna().sort_values("cycle_idx")

    if plot_df.empty:
        print(f"[WARN] No non-null data for: {y_col}")
        return

    plt.figure(figsize=(11, 5))
    plt.plot(plot_df["cycle_idx"], plot_df[y_col], marker=marker, markersize=3)
    finalize_plot(
        title=title,
        xlabel="Cycle index",
        ylabel=ylabel,
        save_name=save_name,
    )


def plot_two_metrics(
    df: pd.DataFrame,
    y1_col: str,
    y2_col: str,
    y1_label: str,
    y2_label: str,
    title: str,
    ylabel: str,
    save_name: str,
):
    ok1 = y1_col in df.columns and df[y1_col].notna().any()
    ok2 = y2_col in df.columns and df[y2_col].notna().any()

    if not ok1 and not ok2:
        print(f"[WARN] Neither {y1_col} nor {y2_col} has plottable data.")
        return

    plt.figure(figsize=(11, 5))

    if ok1:
        d1 = df[["cycle_idx", y1_col]].dropna().sort_values("cycle_idx")
        plt.plot(d1["cycle_idx"], d1[y1_col], marker="o", markersize=3, label=y1_label)

    if ok2:
        d2 = df[["cycle_idx", y2_col]].dropna().sort_values("cycle_idx")
        plt.plot(d2["cycle_idx"], d2[y2_col], marker="s", markersize=3, label=y2_label)

    plt.legend()
    finalize_plot(
        title=title,
        xlabel="Cycle index",
        ylabel=ylabel,
        save_name=save_name,
    )


def plot_multi_columns(
    df: pd.DataFrame,
    cols: list[str],
    title: str,
    ylabel: str,
    save_name: str,
):
    valid_cols = [c for c in cols if c in df.columns and df[c].notna().any()]

    if not valid_cols:
        print(f"[WARN] No valid columns found for plot: {title}")
        return

    plt.figure(figsize=(12, 6))

    for col in valid_cols:
        d = df[["cycle_idx", col]].dropna().sort_values("cycle_idx")
        plt.plot(d["cycle_idx"], d[col], marker="o", markersize=2, label=col)

    plt.legend(loc="best", fontsize=9, ncol=2)
    finalize_plot(
        title=title,
        xlabel="Cycle index",
        ylabel=ylabel,
        save_name=save_name,
    )

# Cell 9
if "final_stage_name" in full_df.columns:
    stage_counts = (
        full_df[["cycle_idx", "final_stage_name"]]
        .dropna()
        .groupby(["cycle_idx", "final_stage_name"])
        .size()
        .reset_index(name="count")
    )

    plt.figure(figsize=(12, 2.5))
    stage_map = {"stage2": 2, "stage3a": 3, "stage3b": 4}
    y = full_df["final_stage_name"].map(stage_map)

    plt.scatter(full_df["cycle_idx"], y, s=30)
    plt.yticks([2, 3, 4], ["stage2", "stage3a", "stage3b"])
    finalize_plot(
        title="Final stage used across cycles",
        xlabel="Cycle index",
        ylabel="Final stage",
        save_name="final_stage_vs_cycle.png",
    )
else:
    print("[WARN] final_stage_name column not found.")



```
